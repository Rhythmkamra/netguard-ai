```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Correct path
df = pd.read_csv('/home/rhythm/Documents/netguard-ai/backend/data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv')

# Column names clean karo
df.columns = df.columns.str.strip()

print("Shape:", df.shape)
print("\nLabel distribution:")
print(df['Label'].value_counts())
print("\nNull values:", df.isnull().sum().sum())
print("Inf values:", (df == float('inf')).sum().sum())
print("Feature names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes.value_counts())

print("\nSample row:")
df.head(2)
plt.figure(figsize=(7, 4))
colors = ['#3B82F6', '#EF4444']
df['Label'].value_counts().plot(kind='bar', color=colors, edgecolor='white')
plt.title('Normal vs DDoS Traffic — CICIDS 2017', fontsize=13)
plt.xlabel('Traffic Type')
plt.ylabel('Packet Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('/home/rhythm/Documents/netguard-ai/notebooks/class_distribution.png')
plt.show()
print("Plot saved!")
```
